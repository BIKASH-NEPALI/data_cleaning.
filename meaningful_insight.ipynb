```python
import json
def load_data(file_name):
    with open(file_name, "r") as f: 
        data=json.load(f)
    return data
```

```python
data=load_data("provided_data.json")
print(data)
```

```python
def cleaning_data(data):
    text_to_num={"one":1, "two": 2 , "threee": 3, "four": 4, "five": 5}
    unique_values=set()
    cleaned_data=[]
    for user in data: 
        #print(user["name"])
        raw_rating=str(user["rating"]).strip()
       # print(raw_rating)
        if (raw_rating in text_to_num):
            raw_rating=text_to_num[raw_rating]
            user["rating"]=raw_rating
        # assigning null values
        raw_age=user.get("age")
        if (raw_age==None):
            user["age"]=None
        # deduplication values 
        if (user["name"] in unique_values):
            continue 
        unique_values.add(user["name"])
        cleaned_data.append(user)
        
    return cleaned_data
```

```python
cleaned_data=cleaning_data(data)
print(cleaned_data)
```

```python
def get_insight(data):
# get the average rating 
    total_rating=0
    for user in data:
        total_rating+=float(user["rating"])
    print(f"Average ratings of cumtomer :{total_rating/len(user)}")
# get percentage with poor ratings 
    poor_ratings =0
    for user in data: 
        if(float(user["rating"])< 3.0):
            poor_ratings+=1
    prcnt=(poor_ratings/len(user))*100
    print(f"total percentage with poor ratings :{prcnt} % ")
```

```python
get_insight(cleaned_data)
```
