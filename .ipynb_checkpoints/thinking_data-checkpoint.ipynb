```python
import json 
def load_data(file_path):
    with open(file_path, "r") as f: 
        data=json.load(f)
    return data
    
    
```

```python
new_data=load_data("provided_data.json")
```

```python
print(new_data)
```

```python
print(type(new_data))
```

```python
# cleaning and structure the data 
def clean_data(data):
    text_to_num={"one": 1, "two": 2, "three": 3, "four" :4, "five":5}
    cleaned_data=[]
    unique_values=set()
    #cleaning rating 
    for el in data : 
        raw_rating=el["rating"]
        if(raw_rating in text_to_num):
            raw_rating=text_to_num[raw_rating]
        el["rating"]=raw_rating
        # cleaning missing values 
        raw_age=el.get("age")
        if(raw_age==None):
            el["age"]=None
        #duplication 
        if(el["name"]  in unique_values):
            continue
        unique_values.add(el["name"]) 
        cleaned_data.append(el)
    return cleaned_data
        
```

```python
new=clean_data(new_data)
print(new)
```
