```python
import json 
```

```python
def load_data(file_name):
    with open(file_name,"r") as f:
        data=json.load(f)
    return data   
    
```

```python
data=load_data("data.json")

print(data)
print(type(data)) # dictionary for python j
```

```python
def clean_data(data):
    for el in data: # print key values of dictionary 
        print(el) 
```

```python
clean_data(data) 
```

```python
def display_users(data):
    for el in data["users"]:
        #print(el)
        #print((el["name"]) is friends of ( el["friends"]) )
        print(f"{el['name']}  is friends of {el['friends']} and user pages are {el['liked_pages']}")
    print("user pages are:\n") 
    for page in data["pages"]:
        print(page['name'])
```

```python
display_users(data) 
```
