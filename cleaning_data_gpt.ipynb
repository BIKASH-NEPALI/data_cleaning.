```python
import json 
```

```python
def load_data(file_name):
    with open(file_name,"r") as f: 
        value=json.load(f) 
        return value 
        
```

```python
data=load_data("data_from_gpt.json")
print(data)
```

```python
def cleaning_data(place_holder):
    for el in place_holder:
        
        el["name"]=el["name"].strip()
        el["email"]=el["email"].strip()
        print(el)
```

```python
cleaning_data(data)
```
